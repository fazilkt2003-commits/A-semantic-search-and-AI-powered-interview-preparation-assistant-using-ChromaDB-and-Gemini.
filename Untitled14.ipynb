## 1. Environment Setup & Dependencies

In this section, we clone the remote datasets repository containing historical interview data and install the required core libraries:
- `llama-index` for data structure utilities.
- `chromadb` as our ultra-fast, local vector database.

```python
import os
import json
import time
import pandas as pd

from tqdm.auto import tqdm
from tqdm.notebook import tqdm
```

## 2. Data Loading & Aggregation

We parse all interview question datasets in `.csv` format from the cloned repository, consolidate them into a unified Pandas DataFrame, and craft a single rich `context` string per record to build high-quality embeddings.

```python
csvs = [csv for csv in os.listdir('/content/datasets/Intervew Questions/') if '.csv' in csv]

df = pd.DataFrame()

for csv in tqdm(csvs):
    df_ = pd.read_csv('/content/datasets/Intervew Questions/' + csv )
    df = pd.concat((df,df_))


df['context'] = "Question " + df['question'] + " asked in the company " + df['company'] + " for the position of " + df['position']

df
```

## 3. Data Cleaning & Deduplication

To ensure optimal RAG performance, we perform deduplication on the combined context texts to eliminate redundancy and drop records missing valid content.

```python
df = df.drop_duplicates(subset=['context'], keep='first')
df = df.dropna(subset=['context'])
```

## 4. Text Distribution Analysis

We analyze and plot the distribution of word counts across our consolidated dataset contexts to verify the length of our context strings and validate that chunking is unnecessary.

```python
df["context"].str.split().str.len().sort_values().reset_index(drop=True).plot(
    kind='line', figsize=(10, 5), title='Word Count per Context'
)
```

## 5. Metadata Preparation & Serialization

We structure our processed dataset records into standard JSON schemas containing text blocks paired with metadata fields (such as dates) before exporting them to a local JSON file.

```python
formatted_data = []

for index, row in tqdm(df.iterrows(), total=df.shape[0]):
    context = row['context']
    date = row['date']
    if isinstance(context, str):
        formatted_data.append({
            'context': context,
            'date': str(date)
        })

output_filepath = 'interview_questions_metadata.json'
with open(output_filepath, 'w', encoding='utf-8') as f:
    json.dump(formatted_data, f, indent=4, ensure_ascii=False)

print(f"Saved {len(formatted_data)} records to {output_filepath}")

# Display the first few elements to verify
import pprint
pprint.pprint(formatted_data[:3])
```

## 6. ChromaDB Collection Setup

We initialize a local, persistent vector client with ChromaDB. To achieve extremely fast, high-quality embeddings, we utilize the `all-MiniLM-L6-v2` SentenceTransformer embedding model mapping straight to the GPU device (`cuda`).

```python
import chromadb
from chromadb.utils.embedding_functions import SentenceTransformerEmbeddingFunction
from chromadb.config import Settings
from tqdm.auto import tqdm

# Initialize client and recreate the collection
client = chromadb.PersistentClient(path="chroma_db", settings=Settings())

old_collection_name = "my_collection"
try:
    client.delete_collection(name=old_collection_name)
    print(f"Deleted collection '{old_collection_name}'.")
except Exception:
    print(f"No existing collection named '{old_collection_name}'.")

ef = SentenceTransformerEmbeddingFunction(model_name="all-MiniLM-L6-v2", device="cuda")
collection = client.get_or_create_collection(name="my_collection", embedding_function=ef)
print("Created collection 'my_collection' with GPU-enabled embedding.")
```

## 7. Vector Database Ingestion

We ingest our 21,013 prepared records into ChromaDB in highly efficient batches of 500 records to prevent memory limits and handle bulk ingestion seamlessly.

```python
# Batch preparation and ingestion
batch_size = 500
num_records = len(formatted_data)

for i in tqdm(range(0, num_records, batch_size), desc="Ingesting documents"):
    batch = formatted_data[i:i + batch_size]

    documents = [item['context'] for item in batch]
    metadatas = [{'date': item['date']} for item in batch]
    ids = [f"doc_{j}" for j in range(i, i + len(batch))]

    collection.add(
        documents=documents,
        metadatas=metadatas,
        ids=ids
    )

print(f"Successfully pushed {collection.count()} items into the collection!")
```

## 8. Query and Verification

We run a baseline search query against our database to verify the indexing and retrieve the top 5 closest matches for Google interview questions.

```python
results = collection.query(
    query_texts=["what are the question asked in google?"],
    n_results=5
)

if results and results.get('documents') and results.get('metadatas'):
    print("Top 5 Interview Questions:")
    for i in range(len(results['documents'][0])):
        document = results['documents'][0][i]
        metadata = results['metadatas'][0][i]

        print(f"\n--- Result {i+1} ---")
        print(f"Date Asked: {metadata.get('date', 'N/A')}")
        print(f"Context: {document}")
else:
    print("No results found or results structure is unexpected.")
```

## 9. Context Assembly for LLM Generation

We fetch the top 10 most relevant results for our target query and package them neatly into a context payload containing metadata boundaries to prevent model hallucination.

```python
query_text = "what are the question asked in google ?"
n_results = 10

results = collection.query(
    query_texts=[query_text],
    n_results=n_results
)

retrieved_articles = []
if results and results.get('documents') and results.get('metadatas'):
    for i in range(len(results['documents'][0])):
        document = results['documents'][0][i]
        metadata = results['metadatas'][0][i]
        retrieved_articles.append({
            'date': metadata.get('date', 'N/A'),
            'chunk': document
        })

if not retrieved_articles:
    print("No relevant articles found in ChromaDB.")
else:
    print(f"Retrieved {len(retrieved_articles)} records. Ready to feed to LLM:")
    import pprint
    pprint.pprint(retrieved_articles)
```

## 10. LLM Client Initialization

We load your secure Gemini API credentials and configure the client to communicate with the state-of-the-art `gemini-3.5-flash` model.

```python
from google import genai
from google.colab import userdata

gemini_client = genai.Client(
    api_key=userdata.get("GOOGLE_API_KEY")
)

gemini_model = "gemini-3.5-flash"
```

## 11. Augmented Response Generation (RAG)

We execute our generation payload with strict instructions ensuring context-bounded answers, highlighting trends, and augmenting explanations with robust model answers for candidates.

```python
# Construct the system instruction and user prompt
system_prompt = (
    "You are a helpful AI assistant. You help students prepare for interviews by providing and "
    "explaining previously asked interview questions. Analyze the retrieved interview questions, "
    "prioritize highlighting any repeated questions or patterns first, and use your knowledge to "
    "augment and explain the answers to these questions to help students prepare thoroughly. "
    "If the answer is not available in the provided articles, say I cannot answer this question based on the provided information."
)

# Format the retrieved questions as context for the model
context_str = "\n\n".join([
    f"Result {idx+1}:\nDate Asked: {art['date']}\nQuestion Details: {art['chunk']}"
    for idx, art in enumerate(retrieved_articles)
])

user_prompt = f"""Here are the retrieved previous interview questions related to the student's query:

{context_str}

Based on the query '{query_text}', please provide a helpful, organized study guide explaining these questions, highlighting any trends/repeated topics first, and providing augmented explanations or model answers for them."""

# Call the Gemini API using the updated model variable
response = gemini_client.models.generate_content(
    model=gemini_model,
    contents=user_prompt,
    config={
        "system_instruction": system_prompt
    }
)

print(response.text)
```
